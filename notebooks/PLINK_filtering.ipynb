```python
import sys
#!pip install XXX --target ./my_custom_packages
sys.path.append('./my_custom_packages')
import polars as pl
import time
#pl.Config.set_tbl_rows(20)
```

```python
humanexom_data = pl.read_csv("GPL18545_humanexome-12v1-1_a.csv", skip_rows = 7, schema_overrides = {"Chr": pl.String})
display(humanexom_data.columns)
display(humanexom_data.head())

merged_dataset = pl.read_parquet("merged_dataset_processed.parquet")
display(merged_dataset.head())

cost_matrix = pl.read_parquet("cost_matrix_processed.parquet")
display(cost_matrix.head())


# #CHROM, POS, ID, REF, ALT, QUAL, FILTER, INFO, FORMAT
```

```python
cost_matrix = cost_matrix.with_columns(
    merged_dataset.select(merged_dataset.columns[0]).rename({"ID_REF": "IlmnID"})
)

cost_matrix = cost_matrix.with_columns(pl.all().cast(pl.String))


gmap = {
    0: "0/0",
    1: "0/1",
    2: "1/1",
    None: "./."
}

cost_matrix = cost_matrix.with_columns(
    pl.col(c).replace(gmap) for c in cost_matrix.columns
)

display(cost_matrix.head())


humanexom_data = humanexom_data.rename({"Name": "ID", "MapInfo": "POS", "Chr": "#CHROM"})

humanexom_data = humanexom_data.with_columns(
    pl.lit(".").alias("QUAL"),
    pl.lit("PASS").alias("FILTER"),
    pl.lit(".").alias("INFO"),
    pl.lit("GT").alias("FORMAT"),
    pl.col("SNP").str.replace_all(r"[\]\[/]", "")
)
humanexom_data = humanexom_data.with_columns(
    pl.col("SNP").str.slice(0, 1).alias("REF"),
    pl.col("SNP").str.slice(1, 1).alias("ALT")

)

humanexom_data = humanexom_data.select(["#CHROM", "POS", "ID", "REF", "ALT", "QUAL", "FILTER", "INFO", "FORMAT", "IlmnID"])

vcf = humanexom_data.join(cost_matrix, on = "IlmnID")

vcf = vcf.select(pl.all().exclude("IlmnID"))



display(vcf.head())
```

```python
filename = "axome_counts.vcf"

with open(filename, "w") as f:
    f.write("##fileformat=VCFv4.2\n")

with open(filename, "a") as f:
    vcf.write_csv(f, separator = "\t")
    
```

```python
%%bash

echo "--- SYSTEM INFO ---"
uname -a
echo "--- FILE INFO ---"
file ./plink2
```

```python
%%bash

#Give the file permission to run
chmod +x plink2

echo "Sorting data"

./plink2 \
--vcf axome_counts.vcf \
--autosome \
--sort-vars \
--make-pgen \
--out axome_counts_plink_data_sorted


# SORT-VARS: sorts the genome in chromosone order
# MAKE-PGEN: saves the file in compressed plink file

echo "Filtering data"
./plink2 \
--pfile axome_counts_plink_data_sorted \
--geno 0.05 \
--mind 0.05 \
--maf 0.01 \
--make-bed \
--out plink_data

# AUTOSOME: Drop the gender chromosone 23 to account for it you can let it guess using impute-sex
# however ml could learning to predict male vs female instead of smoking status
# GENO: Filter when SNP is missing in more than 5% of patients
# MIND: Filters when a genotype is missing more than 5% of SNPs
# MAF: Filters for variant in less than 1% of patients
# MAKE-BED: Converts the data in PLINK's binary format for efficiency
```

```python
%%bash


./plink2 \
--bfile plink_data \
--indep-pairwise 50 5 0.2 \
--out ld_pruning_plink_data

# INDEP-PAIRWISE: 50 SNP window, shift by 5 at a time, and drop when correlated >0.2 with another snp in that window
# OUT: List of remaining snps
```

```python
%%bash

./plink2 \
--bfile plink_data \
--extract ld_pruning_plink_data.prune.in \
--pca 10 \
--out ancestry_pca_plink_data

# To control for genetic background
```
